## O problema

A polarização descreve a orientação do campo elétrico da onda:

* H (Horizontal) → campo elétrico oscila horizontalmente
* V (Vertical) → campo elétrico oscila verticalmente

No RADAR, temos duas polarizações, na transmissão e na recepção:

|Polarização|	Significado|
|--|--|
|VV|	Vertical → Vertical|
|VH|	Vertical → Horizontal|
|HH|	Horizontal → Horizontal|
|HV|	Horizontal → Vertical|

### 1.5. Geometrias RADAR

O imageamento por radar aerotransportado trabalha com duas direções fundamentais:
- Range (alcance): direção transversal à trajetória, associada à distância entre o radar e o alvo.
- Azimute (along-track): direção paralela ao deslocamento da plataforma.

<img src="media/imgs/img7.png" width=800 alt="trasmissão-recepção-polarizações">

O radar voa ao longo de uma linha reta a uma altitude $H$ e observa a Terra sob um ângulo de visada oblíqua (look angle) $\theta_l$. Em vez do ângulo de visada, às vezes é utilizado o ângulo de incidência $\theta_i = (90° – \theta_l)$.

O tamanho da área iluminada na superfície terrestre (a pegada ou footprint do radar) é definido pela largura do feixe da antena $(β)$ e pela distância entre o satélite e o solo $(R)$.

Observe que o feixe do radar é largo na direção de alcance e estreito na direção de azimute.

#### 1.6.2 Critério de Rayleigh

A capacidade de um sistema radar discriminar alvos dependerá da relação do comprimento de onda utilizado do Sistema (λ), altura da rugosidade (h) e ângulo de incidência (θ). Para se ter uma superficie lisa é necessário que:

$ h \lt \frac{\lambda sen(\theta)}{25} \rightarrow $ lisa $\rightarrow $ sem retorno

$ h \gt \frac{\lambda sen(\theta)}{4.4} \rightarrow $ rugosa $\rightarrow $ bom retorno

## 2. Real Aperture Radar (RAR) vs  Synthetic Aperture Radar (SAR)

A principal diferença está na maneira como cada sistema obtém a resolução espacial, especialmente na direção de deslocamento do sensor (azimute).
- RAR — Radar de Abertura Real: **utiliza a abertura física da antena para determinar sua resolução angular**. É uma tecnologia mais simples, historicamente importante no imageamento por radar e atualmente utilizada também no monitoramento de superfícies e estruturas.
- SAR — Radar de Abertura Sintética: **combina sinais captados em diferentes posições do sensor para sintetizar uma antena virtual muito maior**, permitindo formar imagens com resolução espacial elevada, mesmo a grandes distâncias.

<img src="media/imgs/img6.png"  alt="rar-sar">

```python
import numpy as np
import base64
from io import BytesIO
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import HTML, clear_output, display

clear_output(wait=True)
plt.close("all")
if "comparador" in globals():
    for componente in comparador.children:
        componente.close()
    comparador.close()

def comparar_rar_sar(distancia=10, antena=2.0, onda=3.0):

    R, L, lamb = distancia * 1000, antena, onda / 100
    # Resoluções azimutais aproximadas
    rar = lamb * R / L
    sar = L / 2

    fig, ax = plt.subplots(figsize=(7, 2.8))

    barras = ax.barh(
        [0, 0.28],
        [rar, sar],
        color=["#E69F44", "#3088A5"],
        height=0.12
    )

    ax.set_yticks([0, 0.28], labels=["RAR", "SAR"])
    ax.set_ylim(0.65, -0.37)
    ax.set_xlim(0, max(rar, sar) * 1.22 + 1)
    ax.set_xlabel("Resolução azimutal (m)")
    ax.set_title("Comparação da resolução azimutal: RAR × SAR")

    for barra, valor in zip(barras, [rar, sar]):
        ax.text(
            barra.get_width() + max(rar, sar) * 0.02,
            barra.get_y() + barra.get_height() / 2,
            f"{valor:.2f} m",
            va="center",
            fontsize=11,
            fontweight="bold"
        )

    ax.grid(axis="x", linestyle="--", alpha=0.3)
    ax.set_axisbelow(True)
    plt.tight_layout()
    imagem = BytesIO()
    fig.savefig(imagem, format="png", bbox_inches="tight", dpi=100)
    plt.close(fig)
    imagem_base64 = base64.b64encode(imagem.getvalue()).decode("ascii")
    display(HTML(f'<img src="data:image/png;base64,{imagem_base64}" style="display:block;width:100%;max-width:700px;height:auto;">'))

    print(f"RAR: {rar:.2f} m | SAR: {sar:.2f} m")
    print(f"Razão RAR/SAR: {rar/sar:.1f} vezes")


comparador = widgets.interactive(
    comparar_rar_sar,
    distancia=widgets.IntSlider(
        value=10, min=1, max=30, step=1,
        description="Distância (km):",
        style={"description_width": "initial"}
    ),
    antena=widgets.FloatSlider(
        value=2.0, min=0.5, max=10, step=0.5,
        description="Antena (m):",
        style={"description_width": "initial"}
    ),
    onda=widgets.FloatSlider(
        value=3.0, min=1, max=30, step=1,
        description="Onda (cm):",
        style={"description_width": "initial"}
    )
)

controles = widgets.VBox(
    comparador.children[:-1],
    layout=widgets.Layout(width="300px", min_width="0", flex="0 1 300px", padding="20px 10px 0 0", overflow="hidden")
)
for slider in controles.children:
    slider.layout.width = "100%"

saida = comparador.children[-1]
saida.layout = widgets.Layout(flex="1 1 0%", min_width="0", height="auto", max_height="none", overflow="hidden")
saida.add_class("comparador-rar-sar")
display(HTML("""<style>
.comparador-rar-sar .output_scroll,
.comparador-rar-sar .jp-OutputArea-output {
    height: auto !important; max-height: none !important; overflow: visible !important;
}
.comparador-rar-sar img { max-width: 100% !important; height: auto !important; }
</style>"""))
display(widgets.HBox(
    [controles, saida],
    layout=widgets.Layout(width="auto", max_width="100%", min_width="0", overflow="hidden", align_items="flex-start")
))
```
