## 2. Conceituar RAR

### 2.1. Abertura real
No Real Aperture Radar, a separação angular em azimute depende principalmente do diagrama da antena física. Se $D$ é seu comprimento na direção de voo e $\lambda$ o comprimento de onda, a largura angular é aproximadamente $\beta\approx\lambda/D$ radianos.

### 2.2. Consequência cartográfica
A resolução azimutal aproximada é $\delta_a\approx R\lambda/D$. Para $\lambda=0{,}03$ m, $D=3$ m e $R=10$ km, resulta $\delta_a\approx100$ m. A mesma antena produz resolução pior quando o alcance aumenta. A expressão é uma aproximação de largura de feixe; fatores dependem da definição de resolução e da ponderação da antena.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 7.1–7.2; Anexo C.

## 3. Descrever a geometria RAR

### 3.1. Eixos e ângulos
Azimute é a direção aproximadamente paralela ao deslocamento da plataforma; alcance é a direção de visada transversal. O alcance inclinado $R$ une sensor e alvo. Em terreno plano e geometria local, $R^2=H^2+x^2$, com altura $H$ e alcance no solo $x$.

O ângulo de incidência $\theta$ é medido entre a visada e a normal à superfície de referência; em terreno inclinado, a incidência local pode diferir significativamente. Nessa aproximação, $x=H\tan\theta$.

### 3.2. Separar alvos em alcance
Para um pulso não comprimido de duração $\tau$, $\delta_r\approx c\tau/2$. Em terreno plano, $\delta_g\approx\delta_r/\sin\theta$. Dois alvos podem compartilhar o mesmo alcance inclinado e produzir ambiguidade de posição. A dimensão de um pixel da imagem não é, por si só, a resolução física.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 7.1–7.2; Anexo C.

## 4. Conceituar SAR

### 4.1. Abertura sintética
O Synthetic Aperture Radar combina coerentemente os ecos de um alvo observados em diferentes posições da trajetória. A variação de fase permite simular uma abertura maior que a antena física. Não se trata de apenas somar intensidades: é preciso preservar e modelar a fase.

### 4.2. Coerência e movimento
O histórico de fase depende de $R(t)$. Para propagação monostática de ida e volta, sua parcela geométrica tem módulo $4\pi R(t)/\lambda$, com sinal dependente da convenção do sinal complexo. Erros orbitais ou de movimento prejudicam o foco. Amplitude informa sobre a intensidade de espalhamento; fase também carrega informação geométrica, mas é observada módulo $2\pi$.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 7.1–7.2; Anexo C.

## 5. Descrever a geometria SAR

### 5.1. Resolução e modo de aquisição
Em stripmap ideal, broadside e com processamento da abertura completa, a resolução azimutal é aproximadamente $D/2$, diferentemente do RAR. Spotlight e ScanSAR modificam a relação entre cobertura, tempo de iluminação e resolução; a expressão não deve ser aplicada indistintamente.

### 5.2. Distorções do relevo
Vertentes voltadas para o sensor podem sofrer encurtamento (foreshortening). Em condições mais severas, o topo chega antes da base em alcance e ocorre inversão (layover). Regiões sem iluminação apresentam sombra RADAR. A correção geométrica usa trajetória e modelo de elevação, mas não recupera informação ausente na sombra nem separa automaticamente contribuições sobrepostas em layover.

### 5.3. Informação para geoposicionar
O produto precisa informar tempo, trajetória, parâmetros de alcance e modo de aquisição. As seções 7.1–7.2 e o Anexo C da ISO descrevem o contexto do modelo físico SAR e seus metadados.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 7.1–7.2; Anexo C.

## 6. Conceituar LiDAR

### 6.1. Do pulso à nuvem
LiDAR mede distâncias e direções para estimar pontos tridimensionais. Um pulso pode gerar múltiplos retornos ao interceptar copa, ramos e solo por caminhos diferentes. Isso não significa atravessar material opaco: parte da energia alcança o solo por aberturas na vegetação.

### 6.2. Geoposicionamento
Tempo de emissão/recepção, direção do pulso, posição e atitude da plataforma permitem transformar uma medida local em coordenadas terrestres. Erros de sincronização, montagem e navegação afetam a nuvem inteira. A intensidade depende do alvo e da aquisição e não equivale diretamente à reflectância calibrada.

### 6.3. Escolha do sensor
Para mapear inundação sob nuvens, RADAR pode ser adequado; para descrever fachadas e estruturas finas, uma aquisição LiDAR com densidade e geometria apropriadas pode ser preferível. A decisão precisa explicitar escala, acesso, oclusão e precisão exigida.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 9.1–9.2; Anexo D.
