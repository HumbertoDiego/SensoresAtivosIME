## O problema

Após limpar e registrar a nuvem, precisamos identificar terreno, vegetação e construções. Como transformar geometria local em classes e verificar se o modelo generaliza para outro local?

$$
\boxed{\text{Atributos por ponto}} \rightarrow \boxed{\text{Modelo supervisionado}} \rightarrow \boxed{\text{Classes e controle}}

## 1. Construir atributos e uma legenda semântica

### 1.1. Geometria não é semântica
Uma superfície planar pode ser terreno, telhado ou fachada. Descritores locais resumem forma, enquanto a classe depende também de contexto, altura e finalidade cartográfica. A legenda deve definir o tratamento de vegetação baixa, estruturas temporárias e pontos ambíguos.

### 1.2. Atributos complementares
Altura normalizada, normais, autovalores, densidade, intensidade e retornos podem compor o vetor de atributos. A altura depende da qualidade do MDT; a intensidade depende também da aquisição. Usar atributos de múltiplas escalas permite observar detalhes e contexto, ao custo de processamento e maior risco de redundância.

### 1.3. Consistência
As mesmas unidades, regras de vizinhança e tratamentos de valores inválidos devem valer para treino e aplicação. Um classificador pode aprender diferenças de densidade entre campanhas em vez de diferenças entre objetos.

**Referência deste tópico:** Weinmann et al. (2017), *Geometric features and their relevance for 3D point cloud classification* (DOI: 10.5194/isprs-annals-IV-1-W1-157-2017).

**Verificação de aprendizagem:** Como a escala da vizinhança altera a interpretação de um telhado?

## 2. Comparar modelos e organizar treinamento e inferência

### 2.1. Abordagens
Regras por altura e inclinação fornecem uma referência interpretável. Modelos supervisionados por ponto aprendem relações entre atributos e rótulos; modelos que operam em conjuntos de pontos também podem aprender relações espaciais. Maior complexidade exige mais dados, controle e avaliação.

### 2.2. Divisão espacial
Separar regiões ou levantamentos evita que pontos quase coincidentes apareçam no treino e no teste. Se uma nuvem possui sobreposição entre faixas, dividir por arquivo sem avaliar a sobreposição pode manter vazamento. Hiperparâmetros são escolhidos na validação, mantendo o teste reservado.

### 2.3. Inferência em blocos
Nuvens grandes podem exigir blocos com margem de contexto. Calcular atributos usando apenas o interior sem margem cria artefatos nas bordas. Após inferir, remover duplicações e preservar identificadores para voltar à nuvem original.

**Referência deste tópico:** Weinmann et al. (2017), *Geometric features and their relevance for 3D point cloud classification* (DOI: 10.5194/isprs-annals-IV-1-W1-157-2017).

**Verificação de aprendizagem:** Como evitar vazamento quando duas faixas cobrem a mesma área?

## 3. Avaliar classes e discutir limitações cartográficas

### 3.1. Métricas por classe
Para cada classe, $IoU=TP/(TP+FP+FN)$. Se terreno tem $TP=900$, $FP=50$ e $FN=100$, $IoU\approx0{,}857$. A média entre classes atribui o mesmo peso a cada categoria, enquanto agregações globais podem ser dominadas pela classe mais frequente.

### 3.2. Além da contagem de pontos
Regiões densas têm maior peso em métricas por ponto. Complementar com avaliação por área, objeto ou região, conforme o produto. Para MDT, verificar também os erros de elevação resultantes: poucos falsos pontos de terreno sobre telhados podem produzir grandes efeitos locais.

### 3.3. Diagnóstico
Inspecionar confusões em perfis, limites de objetos e áreas de baixa densidade. Comparar desempenho por campanha ou tipo de ambiente revela limites de transferência. Probabilidade alta não garante correção quando o dado de aplicação difere do treinamento.

### 3.4. Síntese para a VF
Justificar a escolha RADAR/LiDAR, distinguir resolução e amostragem, explicar a fase interferométrica e rastrear como erros de aquisição, filtragem e registro afetam uma classificação. A avaliação final deve reunir interpretação física e controle de qualidade.

**Referência deste tópico:** scikit-learn, *Metrics and scoring*.

**Verificação de aprendizagem:** Compare avaliação por ponto e por área em uma nuvem de densidade irregular.
