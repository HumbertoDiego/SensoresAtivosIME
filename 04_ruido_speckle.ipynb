## O problema

Uma região homogênea apresenta pixels claros e escuros. Como reduzir essa variabilidade sem apagar estradas estreitas, bordas e alvos de interesse?

$$
\boxed{\text{Interferência coerente}} \rightarrow \boxed{\text{Imagem granular}} \rightarrow \boxed{\text{Filtro e avaliação}}
$$

## 1. Explicar a origem física e o modelo estatístico do speckle

### 1.1. Interferência na célula de resolução
O sinal complexo soma contribuições de muitos espalhadores. Pequenas diferenças de percurso mudam as fases relativas, produzindo interferência construtiva ou destrutiva. Speckle é uma propriedade do imageamento coerente e depende da cena, da geometria e da resolução.

### 1.2. Modelo multiplicativo
Para intensidade detectada, um modelo ideal é $I=XS$, onde $X$ é a intensidade média de interesse e $S$ tem média unitária. Se $X$ é constante e $S$ é exponencial no caso single-look plenamente desenvolvido, $E[I]=X$ e $\operatorname{Var}(I)=X^2$. Ruído térmico é outro fenômeno, frequentemente modelado como contribuição aditiva antes ou após etapas específicas.

### 1.3. Limites do modelo
Poucos espalhadores dominantes, alvos urbanos e texturas naturais podem violar as hipóteses homogêneas. Uma variação espacial não deve ser chamada de ruído apenas por parecer granular.

## 2. Comparar filtros espaciais e adaptativos

### 2.1. Filtros de janela
A média local reduz variação, mas mistura classes nas bordas. A mediana é robusta a valores extremos, embora não seja um estimador universalmente adequado ao speckle multiplicativo. A escolha da janela define a escala de informação preservada.

### 2.2. Ideia de um filtro adaptativo
Um estimador do tipo Lee pode ser interpretado como $\hat X=\mu_w+k(I-\mu_w)$, em que $\mu_w$ é a média local. O peso $k$ depende da relação entre variância do sinal e variância observada, conforme o modelo adotado. Em áreas homogêneas, o peso tende a favorecer a média; em regiões com estrutura, preserva-se mais do pixel central.

### 2.3. Escolha dos parâmetros
Janelas maiores fornecem mais amostras e misturam estruturas mais distantes. Filtros direcionais ou não locais procuram selecionar vizinhanças mais compatíveis. Comparar parâmetros exige manter o mesmo recorte e a mesma escala de visualização; ajustar o contraste separadamente pode mascarar perda de detalhe.

**Verificação de aprendizagem:** Que efeito uma janela maior pode ter sobre uma estrada estreita?

## 3. Avaliar homogeneidade, bordas e adequação à aplicação

### 3.1. Número equivalente de looks
Em uma região homogênea de intensidade linear, $ENL\approx\bar I^2/s_I^2$. Se a média é 10 e a variância é 25, $ENL\approx4$. Esse cálculo não deve misturar diferentes coberturas nem ser aplicado diretamente aos valores em dB.

### 3.2. Por que ENL não basta
Um filtro que transforma a região em constante teria ENL muito alto e poderia eliminar detalhes úteis. Examinar perfis transversais, bordas, alvos pontuais e a média radiométrica complementa a avaliação. Comparar também a tarefa final: uma estrada preservada pode ser mais relevante que a máxima suavização.

### 3.3. Desenho de comparação
Fixar áreas homogêneas e áreas com limites; medir média e variância antes/depois; inspecionar estruturas finas; e justificar o parâmetro escolhido. Aplicar a mesma transformação de forma consistente aos dados usados em treinamento e inferência.

**Verificação de aprendizagem:** Por que ENL maior pode coexistir com um produto pior?
