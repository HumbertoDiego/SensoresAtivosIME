## O problema

Um alinhamento inicial aproximado ainda deixa fachadas duplicadas. Como refinar a transformação sem confundir vizinhança espacial com correspondência física correta?

$$
\boxed{\text{Alinhamento inicial}} \rightarrow \boxed{\text{Pares e atualização}} \rightarrow \boxed{\text{Validação independente}}
$$

## 1. Explicar o ciclo de otimização ICP

### 1.1. Correspondências alternadas
ICP parte de uma transformação inicial, transforma a fonte, procura correspondências no alvo, estima uma atualização e repete. No ponto a ponto, o objetivo é reduzir distâncias euclidianas entre pares aceitos. A solução por SVD da aula anterior resolve o subproblema de transformação rígida com pares fixos.

### 1.2. Mínimos locais
Ponto mais próximo não é necessariamente ponto homólogo. Fachadas repetidas, baixa sobreposição e uma posição inicial ruim podem conduzir a um alinhamento incorreto. Pré-alinhamento por feições ou controle ajuda a colocar a solução na região de convergência.

### 1.3. Atualização e parada
Se a atualização $\Delta\mathbf T$ atua na fonte já transformada, $\mathbf T_{nova}=\Delta\mathbf T\mathbf T_{atual}$. Critérios de parada podem usar variação da função objetivo, mudança da transformação e máximo de iterações. Parar não comprova que a solução é correta.

**Referência deste tópico:** Open3D, *ICP registration*.

**Verificação de aprendizagem:** Por que a convergência do ICP não comprova o alinhamento correto?

## 2. Comparar funções objetivo e estratégias robustas

### 2.1. Ponto a ponto
O resíduo é $\mathbf r_i=\mathbf R\mathbf p_i+\mathbf t-\mathbf q_i$. Sua norma penaliza erros nas três direções. Densidades muito diferentes e correspondências incorretas podem enviesar a estimativa.

### 2.2. Ponto a plano
Com normal unitária $\mathbf n_i$ no alvo, minimize $\sum_i[\mathbf n_i^T(\mathbf R\mathbf p_i+\mathbf t-\mathbf q_i)]^2$. A distância normal aproxima bem superfícies locais, mas depende de normais confiáveis. Em um único plano, deslocamentos tangenciais podem permanecer mal determinados.

### 2.3. Rejeição e multiescala
Um limiar de distância limita pares aceitos; perdas robustas reduzem a influência de resíduos altos. Começar em resolução mais grosseira e refinar pode reduzir custo e ampliar a tolerância inicial. O limiar precisa acompanhar ruído, amostragem e escala, sem descartar quase todos os pares.

**Referência deste tópico:** Open3D, *ICP registration*.

**Verificação de aprendizagem:** Quais movimentos ficam mal determinados em uma única superfície plana?

## 3. Validar registro e preparar a prática

### 3.1. Resíduo e cobertura
RMSE sobre pares aceitos mede ajuste condicionado ao pareamento e ao limiar. Um RMSE muito baixo pode refletir poucos pares convenientes. Relatar também sobreposição, número de correspondências e distribuição espacial dos resíduos; verificar a definição de fitness da biblioteca utilizada.

### 3.2. Teste independente
Reservar pontos ou superfícies de controle. Inspecionar cortes em diferentes orientações e locais. Um cenário contendo apenas superfícies paralelas pode deixar componentes da transformação pouco observáveis, mesmo quando o resultado visual parece bom.

### 3.3. Mudança real versus erro
Vegetação móvel e objetos removidos não devem necessariamente entrar no registro. Caso contrário, a transformação pode absorver mudança real. Preferir estruturas estáveis e avaliar separadamente as regiões modificadas.

### 3.4. Preparação para a 2ª VE prática
Entregar convenção fonte/alvo, transformação homogênea, parâmetros de vizinhança e rejeição, estado inicial/final e controle independente. Demonstrar o efeito de uma inicialização ruim e justificar o alinhamento escolhido.

**Referência deste tópico:** Open3D, *ICP registration*.

**Verificação de aprendizagem:** Como um limiar muito pequeno pode produzir RMSE baixo e registro ruim?
