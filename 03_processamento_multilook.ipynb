## O problema

Uma imagem SAR contém valores digitais, geometria inclinada e textura granular. Quais operações permitem comparar o retroespalhamento entre áreas e escolher uma escala adequada ao mapa?

$$
\boxed{\text{Produto e metadados}} \rightarrow \boxed{\text{Calibração e multilook}} \rightarrow \boxed{\text{Correção para o terreno}}
$$

## 1. Interpretar produtos e planejar o processamento

### 1.1. Complexo e intensidade
Um pixel complexo pode ser escrito $z=a+ib=Ae^{i\phi}$. A amplitude é $A=\sqrt{a^2+b^2}$ e a intensidade $I=|z|^2$. Produtos SLC preservam fase; produtos detectados, como GRD, não fornecem a fase complexa original necessária à interferometria.

### 1.2. Ordem e finalidade
Identificar missão, modo, polarização, órbita, datas e processamento anterior é o primeiro passo. Um fluxo de intensidade pode incluir atualização orbital, tratamento de ruídos do instrumento conforme o produto, calibração, multilook ou filtragem e correção de terreno. A ordem específica depende do operador e do produto; não existe uma receita universal para todas as missões.

### 1.3. Rastreabilidade
Registrar área de interesse, parâmetros e versão do processador permite repetir a análise. O valor zero pode representar ausência de dados; tratá-lo como retorno físico muito baixo distorce estatísticas e logaritmos.

## 2. Explicar calibração radiométrica e correção geométrica

### 2.1. Valores comparáveis
Calibrar converte números digitais por meio dos fatores fornecidos no produto. $\beta^0$, $\sigma^0$ e $\gamma^0$ usam diferentes normalizações de área. A escolha deve acompanhar a finalidade e a geometria. Correção de terreno e normalização radiométrica de terreno resolvem problemas distintos.

### 2.2. Escala linear e decibéis
Para potência, $I_{dB}=10\log_{10}I$; para amplitude, $20\log_{10}A$. Uma potência linear de $0{,}01$ corresponde a $-20$ dB. Médias físicas de potência devem ser calculadas em escala linear, pois a média de logaritmos não é o logaritmo da média.

### 2.3. Geocodificação
Órbita, tempos, geometria de alcance e modelo de elevação permitem relacionar pixels ao terreno. O tamanho de saída deve ser compatível com a resolução de entrada. Reprojetar altera a grade, mas não cria informação nem remove sombra e layover.

**Referência deste tópico:** NASA/SERVIR (2019), *The SAR Handbook*, cap. 2.

**Verificação de aprendizagem:** Compare a média linear de 0,01 e 0,1 com a média dos respectivos valores em dB.

## 3. Analisar multilook e o compromisso entre variância e detalhe

### 3.1. Integração não coerente
No multilook, intensidades de looks idealmente independentes são combinadas: $\bar I=L^{-1}\sum_{\ell=1}^{L}I_\ell$. Não se deve confundir essa média com a soma coerente de sinais complexos utilizada no foco SAR.

### 3.2. Dedução do ganho estatístico
Se cada look tem média $\mu$ e variância $\sigma^2$ e os looks são independentes, $E[\bar I]=\mu$ e $\operatorname{Var}(\bar I)=\sigma^2/L$. Para speckle plenamente desenvolvido de intensidade, $\sigma^2=\mu^2$ e o coeficiente de variação é $1/\sqrt L$. Correlação entre looks reduz o ganho efetivo.

### 3.3. Decisão de escala
Combinar looks reduz a variabilidade, mas geralmente alarga o suporte espacial. Quatro looks independentes reduzem o desvio padrão à metade; não significam necessariamente pixels quatro vezes maiores em cada direção. A configuração em alcance e azimute deve considerar suas resoluções no solo.

**Referência deste tópico:** NASA ARSET (2020), *SAR, parte 1*.

**Verificação de aprendizagem:** Qual é a redução ideal do desvio padrão com nove looks independentes?
