## 1. Definir classes e atributos SAR

### 1.1. Do espalhamento à classe
Uma classe temática descreve uma categoria de interesse; o sensor registra resposta eletromagnética. A relação não é unívoca: diferentes classes podem compartilhar intensidade, e uma classe pode variar com umidade, rugosidade e orientação. Por isso, a legenda deve definir o que conta como água, vegetação ou construção e como lidar com áreas ambíguas.

### 1.2. Vetor de atributos
Uma amostra pode reunir intensidades por polarização, razões, estatísticas de textura e medidas multitemporais. Uma razão de potências deve usar valores lineares; em dB, sua expressão equivalente é uma diferença. Texturas exigem janela, quantização e escala documentadas.

### 1.3. Referência temática
Rótulos precisam de fonte, data e grau de confiança. Uma referência de outra estação pode discordar por mudança real. Pixels mistos e bordas podem exigir uma política específica de amostragem.

**Referência deste tópico:** NASA/SERVIR (2019), *The SAR Handbook*, cap. 2.

**Verificação de aprendizagem:** Que evidências distinguiriam água de sombra RADAR?

## 2. Separar treinamento, validação e teste sem vazamento

### 2.1. Papéis dos conjuntos
Treinamento estima parâmetros do modelo; validação orienta hiperparâmetros e seleção; teste estima desempenho após congelar as escolhas. O teste não pode orientar repetidamente decisões de desenvolvimento.

### 2.2. Dependência espacial
Pixels vizinhos compartilham contexto e podem ter atributos derivados da mesma janela. Uma divisão aleatória de pixels pode colocar partes quase idênticas no treino e no teste. Separar por polígonos, blocos, cenas ou datas, conforme o uso pretendido, torna a avaliação mais representativa. Faixas de separação podem ser necessárias quando atributos usam vizinhança.

### 2.3. Pré-processamento aprendido
Normalização, seleção de atributos e balanceamento devem ser ajustados apenas no treinamento de cada divisão. Comparar um limiar simples com um classificador supervisionado ajuda a verificar se a complexidade oferece ganho real.

**Referência deste tópico:** scikit-learn, *Metrics and scoring*.

**Verificação de aprendizagem:** Desenhe uma divisão espacial para avaliar generalização a outra região.

## 3. Interpretar métricas e executar inferência

### 3.1. Matriz de confusão
Adotando linhas como referência e colunas como predição, a diagonal contém acertos. Para uma classe, precisão $=TP/(TP+FP)$ e revocação $=TP/(TP+FN)$. $F_1=2PR/(P+R)$ combina as duas. A convenção da matriz deve acompanhar sua apresentação.

### 3.2. Cálculo ilustrativo
Se o mapa contém 80 verdadeiros positivos, 20 falsos positivos e 40 falsos negativos para água, $P=0{,}80$, $R\approx0{,}67$ e $F_1\approx0{,}73$. Uma boa precisão pode coexistir com muitas omissões. Classes raras exigem resultados por classe, além da acurácia global.

### 3.3. Inferência e entrega
Aplicar ao restante da cena a mesma cadeia de atributos e o modelo congelado. Guardar máscara de validade e, quando disponível, probabilidades ou medida de confiança. A qualidade do mapa depende também da representatividade da amostra e da distribuição espacial dos erros.

### 3.4. Preparação para a 1ª VE prática
Produzir um mapa, registrar a divisão espacial, comparar alternativas pela validação e apresentar a matriz de confusão do teste reservado. Discutir exemplos de comissão e omissão com a geometria SAR.

**Referência deste tópico:** scikit-learn, *Metrics and scoring*.
