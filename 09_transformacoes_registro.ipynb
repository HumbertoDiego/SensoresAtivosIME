## O problema

Duas nuvens do mesmo local apresentam paredes duplicadas. A diferença é uma mudança real, um sistema de coordenadas incompatível ou uma transformação entre levantamentos?

## 1. Distinguir mudança de referencial e registro geométrico

### 1.1. Coordenadas precisam de contexto
Uma tripla XYZ só é interpretável com unidade, eixos, origem, datum e época quando pertinente. Reprojeção cartográfica usa uma relação definida entre sistemas; registro estima uma relação geométrica com observações comuns. Aplicar ICP não corrige automaticamente uma confusão entre altura elipsoidal e ortométrica.

### 1.2. Modelo rígido
Adotando vetores coluna e transformação da fonte para o alvo, $\mathbf q=\mathbf R\mathbf p+\mathbf t$. A matriz deve satisfazer $\mathbf R^T\mathbf R=\mathbf I$ e $\det\mathbf R=+1$. São seis graus de liberdade: três de rotação e três de translação.

### 1.3. Escolha do modelo
Uma similaridade acrescenta escala: $\mathbf q=s\mathbf R\mathbf p+\mathbf t$. Um modelo afim permite deformações mais gerais, mas pode absorver erros e distorcer distâncias. A complexidade deve ser justificada pela origem dos dados, e não apenas pela redução do resíduo.

**Referência deste tópico:** Umeyama (1991), *Least-Squares Estimation of Transformation Parameters Between Two Point Patterns*, IEEE TPAMI, 13(4), 376–380 (DOI: 10.1109/34.88573).

**Verificação de aprendizagem:** Quando escolher transformação rígida em vez de similaridade?

## 2. Compor e inverter transformações entre espaços

### 2.1. Coordenadas homogêneas
$$\begin{bmatrix}\mathbf q\\1\end{bmatrix}=\mathbf T\begin{bmatrix}\mathbf p\\1\end{bmatrix},\qquad \mathbf T=\begin{bmatrix}\mathbf R&\mathbf t\\0&1\end{bmatrix}.$$
Se $\mathbf T_{BA}$ leva de A para B e $\mathbf T_{CB}$ de B para C, então $\mathbf T_{CA}=\mathbf T_{CB}\mathbf T_{BA}$. A matriz à direita atua primeiro. A ordem é essencial porque rotações e translações, em geral, não comutam.

### 2.2. Transformação inversa
Para transformação rígida, $\mathbf T^{-1}=[\mathbf R^T,-\mathbf R^T\mathbf t;0,1]$. Transpor a matriz homogênea inteira não produz a inversa. Conferir uma ida e volta sobre pontos conhecidos ajuda a detectar erros de convenção.

### 2.3. Exemplo
Uma translação de 10 m em X seguida de rotação de 90° em Z leva a origem a $(0,10,0)$. Aplicar a rotação primeiro e a translação depois leva a $(10,0,0)$. O exemplo adota rotação ativa positiva e regra da mão direita.

**Referência deste tópico:** Umeyama (1991), *Least-Squares Estimation of Transformation Parameters Between Two Point Patterns*, IEEE TPAMI, 13(4), 376–380 (DOI: 10.1109/34.88573).

**Verificação de aprendizagem:** Demonstre a inversa por multiplicação de matrizes.

## 3. Estimar transformação rígida por pontos homólogos

### 3.1. Correspondências
Pontos homólogos representam a mesma feição física. Para uma solução rígida 3D ideal, três pares não colineares são o mínimo; mais pares bem distribuídos permitem redundância e diagnóstico. Identificar o mesmo detalhe com precisão é mais importante que apenas acumular pares.

### 3.2. Solução por mínimos quadrados
Minimize $\sum_i\|\mathbf q_i-(\mathbf R\mathbf p_i+\mathbf t)\|^2$. Calcule centroides e centralize as coordenadas. Com $\mathbf H=\sum_i\mathbf p_i'\mathbf q_i'^T=\mathbf U\mathbf\Sigma\mathbf V^T$, use
$$\mathbf R=\mathbf V\operatorname{diag}(1,1,\det(\mathbf V\mathbf U^T))\mathbf U^T,$$
e $\mathbf t=\bar{\mathbf q}-\mathbf R\bar{\mathbf p}$. A correção do determinante evita reflexão.

### 3.3. Qualidade externa
Inspecionar resíduos vetoriais e validar com feições reservadas. Um erro pequeno nos pares usados no ajuste não garante qualidade fora da região que eles cobrem. Erros grosseiros de correspondência exigem identificação ou estimação robusta antes do refinamento.

**Referência deste tópico:** Umeyama (1991), *Least-Squares Estimation of Transformation Parameters Between Two Point Patterns*, IEEE TPAMI, 13(4), 376–380 (DOI: 10.1109/34.88573).

**Verificação de aprendizagem:** Por que três pares colineares não determinam uma rotação 3D única?
