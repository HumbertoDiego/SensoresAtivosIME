## O problema

Uma nuvem parece detalhada, mas apresenta faixas deslocadas e áreas sem pontos no solo. Quais medidas produziram os pontos e o que elas permitem afirmar sobre o terreno?

$$
\boxed{\text{Tempo e direção do pulso}} \rightarrow \boxed{\text{Posição e atitude}} \rightarrow \boxed{\text{Ponto no referencial terrestre}}
$$

## 1. Descrever aquisição LiDAR e retornos

### 1.1. Distância e amostragem
Para tempo de voo, $\rho=v\Delta t/2$. Com $\Delta t=2\,\mu s$ e $v\approx c$, a distância é aproximadamente 300 m. A medida depende da identificação temporal do eco, do meio e do instrumento.

### 1.2. Retornos discretos e forma de onda
O receptor pode registrar retornos discretos ou amostras da forma de onda. Primeiro e último retorno descrevem a ordem dos ecos detectados, não rótulos semânticos. O último retorno pode vir de um telhado ou da copa se o solo não for alcançado.

### 1.3. Densidade e cobertura
Densidade depende da frequência de pulsos, velocidade, padrão de varredura e sobreposição. Espaçamento médio não descreve oclusões nem garante detectar um objeto estreito. A seção 9.1 da ISO apresenta características e tipos de LiDAR.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 6 e 9; Anexo D.

**Verificação de aprendizagem:** Explique a diferença entre densidade média e cobertura efetiva.

## 2. Explicar georreferenciamento e fontes de incerteza

### 2.1. Cadeia de referenciais
Uma expressão didática é
$$\mathbf p^E=\mathbf x^E(t)+\mathbf R_B^E(t)\left[\mathbf l^B+\mathbf R_S^B\rho\mathbf u^S\right].$$
$E$ é o referencial terrestre, $B$ o corpo da plataforma e $S$ o sensor; $\mathbf l^B$ é o braço de alavanca; $\mathbf R_S^B$ descreve o alinhamento; e $\mathbf u^S$ é a direção unitária do pulso. A origem escolhida para $\mathbf x^E$ deve ser consistente com o braço de alavanca.

### 2.2. Sincronização e calibração
Posição GNSS, atitude inercial e pulso devem compartilhar uma escala temporal conhecida. Erros de boresight podem gerar discrepâncias sistemáticas entre faixas. A ISO, seção 9.2 e Anexo D, organiza metadados para geoposicionamento; a equação acima é uma síntese didática, não uma transcrição da norma.

### 2.3. Propagação de erros
Para pequenos erros angulares, o erro transversal cresce aproximadamente como $\rho\delta\alpha$. A 1000 m, um erro de $0{,}001$ rad pode produzir cerca de 1 m. Isso explica por que uma boa medida de distância não basta para uma boa coordenada.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 6 e 9; Anexo D.

**Verificação de aprendizagem:** Estime o erro transversal de 0,01° a 500 m.

## 3. Relacionar aplicações, produtos e controle de qualidade

### 3.1. Superfície, terreno e altura normalizada
MDS representa a superfície visível; MDT representa o terreno, após identificar pontos adequados e modelar lacunas. Uma altura normalizada pode ser calculada por $h_n=z-z_{MDT}(x,y)$. Sua qualidade depende do MDT e do referencial vertical.

### 3.2. Aplicações cartográficas
Relevo, drenagem, vegetação, redes elétricas e estruturas urbanas exigem densidades e geometrias diferentes. Pontos em coordenadas projetadas e alturas ortométricas não podem ser misturados inadvertidamente com coordenadas geocêntricas e alturas elipsoidais.

### 3.3. Inspeção antes do processamento
Verificar unidade, sistema de referência, distribuição de densidade, limites das faixas e pontos de controle. Inspecionar perfis ajuda a identificar superfícies duplicadas. Registrar formato, escala e offset das coordenadas evita interpretações erradas dos valores armazenados.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 6 e 9; Anexo D.

**Verificação de aprendizagem:** Que inspeção ajuda a detectar erro de alinhamento entre faixas?
