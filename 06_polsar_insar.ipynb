## O problema

Duas áreas têm intensidades semelhantes, mas estruturas diferentes; duas imagens da mesma área apresentam diferenças de fase. Como explorar essas duas fontes de informação sem confundi-las?

$$
\boxed{\text{Polarização: estrutura}} \rightarrow \boxed{\text{Fase: caminho percorrido}} \rightarrow \boxed{\text{Interpretação física}}
$$

## 1. Explicar polarimetria e mecanismos de espalhamento

### 1.1. Canais polarimétricos
Polarização descreve a orientação do campo elétrico. Canais HH, HV, VH e VV combinam estados de transmissão e recepção; é preciso verificar a convenção dos índices do produto. No arranjo matricial com linha de recepção e coluna de transmissão, $\mathbf E_r\propto\mathbf S\mathbf E_t$, com $\mathbf S$ complexa de dimensão $2\times2$.

### 1.2. Informação complementar
Espalhamento superficial, dupla reflexão e volume são modelos físicos úteis. Nenhum canal isolado identifica universalmente uma classe. Fases relativas e correlações entre canais acrescentam informação à intensidade; produtos dual-pol não fornecem toda a matriz de um produto quad-pol.

### 1.3. Calibração
Comparações polarimétricas pressupõem correção dos desequilíbrios entre canais e controle da contaminação cruzada. Uma interpretação física exige observar geometria, frequência e limitações do produto.

**Referência deste tópico:** NASA/JPL, *SAR Polarimetry*.

## 2. Relacionar fase interferométrica e geometria

### 2.1. Interferograma
Para dois SLC corregistrados, $q=z_1z_2^*$ e $\phi=\arg(q)$. A ordem do produto define o sinal. Em uma convenção monostática de duas passagens, a diferença de fase geométrica é proporcional a $4\pi\Delta R/\lambda$; o sinal depende de como se define $\Delta R$.

### 2.2. Linha de base e coerência
A separação entre posições de observação é a linha de base. Sua componente perpendicular influencia a sensibilidade topográfica. A coerência estimada numa janela é
$$\hat\gamma=\frac{|\sum z_1z_2^*|}{\sqrt{\sum|z_1|^2\sum|z_2|^2}}.$$
Mudanças nos espalhadores, geometria, ruído e erros de registro reduzem a coerência.

### 2.3. Ambiguidade de fase
A fase observada é enrolada em um intervalo de comprimento $2\pi$. Desenrolar busca estimar múltiplos inteiros desse período com hipóteses de continuidade. Baixa coerência e descontinuidades podem tornar a solução ambígua.

**Referência deste tópico:** ISO/TS 19130-2:2014, *Geographic information — Imagery sensor models for geopositioning — Part 2: SAR, InSAR, lidar and sonar*; seções 8.1–8.3.

## 3. Descrever processamento InSAR e geração de MDT

### 3.1. Cadeia conceitual
Selecionar SLC compatíveis; estimar órbita e geometria; corregistrar; formar interferograma; remover componentes previstas; avaliar coerência; filtrar quando adequado; desenrolar fase; inverter para altura ou deslocamento; e geocodificar. Produtos detectados sem fase não substituem SLC.

### 3.2. Altura e deslocamento
A fase pode conter topografia, deslocamento na linha de visada, atmosfera, erros orbitais e ruído. Uma inversão topográfica usa geometria e linha de base; uma análise de deformação precisa remover ou estimar a contribuição topográfica. Uma única geometria fornece deslocamento projetado, não o vetor 3D completo.

### 3.3. MDT ou superfície de espalhamento?
A elevação recuperada representa o centro de fase dominante, que pode estar na vegetação ou em estruturas. Chamá-la de MDT de terreno nu exige validação e tratamento adicional. Em áreas sem coerência, a interpolação não equivale a uma medição.

### 3.4. Exemplo dimensional
Com $\lambda=0{,}056$ m, uma variação de caminho monostático equivalente a $\lambda/2=0{,}028$ m produz um ciclo completo de fase. Isso ilustra sensibilidade, e não garante precisão milimétrica em qualquer cena.
